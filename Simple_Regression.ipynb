```python
x=[5, 3, -1, 2, 6, 0, 7]
y=[14, 6, -5.5, 3.5, 18, -1, 19.5]
print(x,y)
print(len(x), len(y))
```

```python
X, Y, XiYi, Xi2 = 0, 0, 0, 0
n=len(x)
for i in range(n):
  X= X + x[i]
  Y= Y + y[i]
  XiYi= XiYi + x[i] * y[i]
  Xi2 = Xi2 + x[i] * x[i]
print(n, X, Y, XiYi, Xi2)
```

```python
w1= (n * XiYi - X * Y) / (n * Xi2 - X * X)
w0= (Y - w1 * X) / n
print(w1, w0)
```
